# Dimensionality Reduction Assignments

## 1. Applying PCA

```python
# 1. Read in the student grades data set
import pandas as pd
df = pd.read_csv("../Data/student_grades.csv")
df.head()
```

```python
# 2. Drop the first column with student_id
df_grades = df.drop(columns=["student_id"])
df_grades.head()
```

```python
# 3. Center the data
df_grades=df_grades-df_grades.mean()
df_grades.head()
```

```python
# 4. Fit a PCA model with 2 components
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
pca.fit(df_grades)
```

```python
# 5. View and interpret the explained variance ratios
pca.explained_variance_ratio_
```

## 2. Interpreting PCA 

```python
# 1. View and interpret the components of the PCA model
pca.components_
#math,science,cs,band, english, history, spanish, physed
#pc1 doing well overall
#pc2 math,science,cs,physed vs english, history, spanish
pd.DataFrame(pca.components_,columns=["math","science","cs","band","english","history","spanish","physed"])
```

```python
# 2. Plot the students on a scatter plot with the x-axis as PC 1 and the y-axis as PC 2
import seaborn as sns
import matplotlib.pyplot as plt
pca.transform(df_grades)
df_grades_pca  = pd.DataFrame(pca.transform(df_grades),columns=["pc1","pc2"])
sns.scatterplot(x="pc1",y="pc2",data=df_grades_pca)
plt.xlabel("overall grades")
plt.ylabel("math,science,cs vs enlish, history spanish")

```

```python
# most students prefer either science or humanities, few are doing well across all subjects. 
```
